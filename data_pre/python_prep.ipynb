### Find Duplicate Elements in a List

```python
nums = [1,2,3,4,2,5,3,6]
no_dub = set()
dubli = set()
for i in nums:
    if i not in dubli:
        dubli.add(i)
    else:
        no_dub.add(i)
print(list(dubli))
print(list(no_dub))
```

### Count Frequency of Elements in a List

```python
nums = [1,2,2,3,3,3,4]
freq = {}
for n in nums:
    freq[n] = freq.get(n, 0) + 1
print(freq)
```

### Find the Second Largest Number

```python
nums = [10, 20, 4, 45, 99]
num1 = list(set(nums))
num1.sort(reverse=True)
print(num1[1])
print(num1)
```

### Reverse a String

```python
text = "dataengineering"
reverserd_text = text[::-1]
print(reverserd_text)
```

### Check if a String is a Palindrome

```python
word = "level"
reverserd_word = word[::-1]
print(reverserd_word)
def is_palindrome(word):
    return word == word[::-1]
word = "level"
print(is_palindrome(word))
```
